## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Research Question

Which pages should be reviewed first for content refresh based on observable search and engagement signals?

Unit of analysis

One webpage (one content item).

Output

A ranked refresh score or probability that a page is at risk of declining.

Decision

Which pages should the content team review first?

Action

Review the page and decide whether to update content, improve metadata, expand the article, or continue monitoring it.

Cost of a wrong recommendation

False positive

Editors spend time reviewing a page that did not actually need work.

False negative

A page that truly needs attention is ignored, leading to lost search traffic.

Why ML helps

Many signals such as impressions, CTR, position, freshness, engagement, and content age interact together. Machine learning can learn these relationships better than a fixed scoring rule and produce a ranked review queue.
